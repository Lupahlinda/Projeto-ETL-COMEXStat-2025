Este notebook é responsável por gerar os modelos de dados finais do pipeline ETL:

Execute primeiro os notebooks:

## Importações e Configurações

```python
import pandas as pd
import os
import sys
from datetime import datetime

# Configurações
OUTPUT_DIR = '../data/output'
DADOS_TRANSFORMADOS = os.path.join(OUTPUT_DIR, 'dados_transformados.csv')

# Adicionar diretório src ao path
sys.path.append('../src')

# Importar funções do módulo load
from load import gerar_sql_mysql_star_schema, executar_load_completo

print("✓ Bibliotecas importadas com sucesso!")
```

## Carregar Dados da Etapa Anterior

```python
if os.path.exists(DADOS_TRANSFORMADOS):
    df = pd.read_csv(DADOS_TRANSFORMADOS)
    print(f"✓ Dados carregados: {len(df):,} registros")
    print(f"  Colunas: {len(df.columns)}")
else:
    raise FileNotFoundError(f"Execute primeiro o notebook 02_Transform.ipynb!")
```

## Verificar Estrutura dos Dados

```python
print("=== VERIFICANDO ESTRUTURA DOS DADOS ===")
print(f"Total de registros: {len(df):,}")
print(f"Total de colunas: {len(df.columns)}")
print(f"\nColunas disponíveis:")
for i, col in enumerate(df.columns, 1):
    print(f"  {i}. {col}")
```

## Etapa 1: Geração do Script SQL

```python
print("=== ETAPA 1: GERAÇÃO DO SCRIPT SQL ===")
print("Gerando script SQL MySQL com Star Schema...")

# Chamar função para gerar SQL
gerar_sql_mysql_star_schema(df)

print("✓ Script SQL gerado com sucesso!")
```

## Verificar Arquivo SQL Gerado

```python
sql_file = os.path.join(OUTPUT_DIR, 'comexstat_mysql_schema.sql')

if os.path.exists(sql_file):
    tamanho = os.path.getsize(sql_file) / (1024)  # KB
    print(f"✓ Arquivo SQL gerado: {sql_file}")
    print(f"  Tamanho: {tamanho:.2f} KB")
else:
    print("✗ Arquivo SQL não encontrado!")
```

```python
# Mostrar primeiras linhas do arquivo SQL
if os.path.exists(sql_file):
    with open(sql_file, 'r', encoding='utf-8') as f:
        linhas = f.readlines()[:30]
        print("=== PRIMEIRAS 30 LINHAS DO SCRIPT SQL ===")
        for i, linha in enumerate(linhas, 1):
            print(f"{i:3d}: {linha}")
```

## Etapa 2: Geração da Documentação HTML

```python
print("=== ETAPA 2: GERAÇÃO DA DOCUMENTAÇÃO HTML ===")
print("Gerando documentação HTML com modelos de dados...")

# Chamar função completa de load (gera SQL e HTML)
executar_load_completo(df, None)  # None para não criar dados_finais.csv

print("✓ Documentação HTML gerada com sucesso!")
```

## Verificar Arquivo HTML Gerado

```python
html_file = os.path.join(OUTPUT_DIR, 'modelo_conceitual.html')

if os.path.exists(html_file):
    tamanho = os.path.getsize(html_file) / (1024)  # KB
    print(f"✓ Arquivo HTML gerado: {html_file}")
    print(f"  Tamanho: {tamanho:.2f} KB")
else:
    print("✗ Arquivo HTML não encontrado!")
```

## Verificar Todos os Arquivos Gerados

```python
print("=== ARQUIVOS GERADOS NA ETAPA LOAD ===")
arquivos_gerados = [
    'comexstat_mysql_schema.sql',
    'modelo_conceitual.html',
    'dados_carregados.csv',
    'dados_transformados.csv',
    'dados_com_previsoes.csv',
    'grafico_previsao_completo.png'
]

for arquivo in arquivos_gerados:
    caminho = os.path.join(OUTPUT_DIR, arquivo)
    if os.path.exists(caminho):
        tamanho = os.path.getsize(caminho) / (1024*1024)  # MB
        print(f"  ✓ {arquivo}: {tamanho:.2f} MB")
    else:
        print(f"  ✗ {arquivo}: NÃO ENCONTRADO")
```
