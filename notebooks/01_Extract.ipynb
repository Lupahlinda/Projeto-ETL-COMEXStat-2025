# 01 - Extract: Carregamento de Dados

## Objetivo
Este notebook é responsável por carregar os dados brutos de exportação brasileira do COMEX Stat 2025.

### Etapas do Processo:
1. Verificar existência dos arquivos de entrada
2. Carregar dados de exportações
3. Realizar análise exploratória inicial
4. Salvar dados processados para a próxima etapa

## Importações e Configurações

```python
import pandas as pd
import os
import requests

# Configurações
INPUT_DIR = '../data/input'
OUTPUT_DIR = '../data/output'
DICT_DIR = '../data/dictionaries'

# Arquivos de entrada
ARQUIVO_EXPORTACOES = os.path.join(INPUT_DIR, 'Exportacoes_reduzidos.csv')

print("✓ Bibliotecas importadas com sucesso!")
```

## Funções de Leitura de Dados

```python
def ler_dados_csv(file_path, delimiter=";"):
    """Lê arquivo CSV e retorna DataFrame."""
    df = pd.read_csv(file_path, sep=delimiter)
    return df

def ler_csv_para_validacao(file_path, delimiter=";"):
    """Lê arquivo CSV para validação."""
    df = pd.read_csv(file_path, sep=delimiter)
    return df

print("✓ Funções de leitura definidas!")
```

## Verificar Arquivos de Entrada

```python
# Verificar se o arquivo principal existe
if os.path.exists(ARQUIVO_EXPORTACOES):
    tamanho = os.path.getsize(ARQUIVO_EXPORTACOES) / (1024*1024)
    print(f"✓ Arquivo de exportações encontrado: {ARQUIVO_EXPORTACOES}")
    print(f"  Tamanho: {tamanho:.2f} MB")
else:
    raise FileNotFoundError(f"Arquivo não encontrado: {ARQUIVO_EXPORTACOES}")
```

## Carregar Dados de Exportações

```python
print("Carregando dados de exportações...")
df = ler_dados_csv(ARQUIVO_EXPORTACOES)

print(f"\n✓ Dados carregados com sucesso!")
print(f"  Total de registros: {len(df):,}")
print(f"  Total de colunas: {len(df.columns)}")
```

## Análise Exploratória Inicial

```python
# Exibir as primeiras linhas
print("=== PRIMEIRAS LINHAS DOS DADOS ===")
df.head()
```

```python
# Informações sobre o dataset
print("=== INFORMAÇÕES DO DATASET ===")
df.info()
```

```python
# Estatísticas descritivas
print("=== ESTATÍSTICAS DESCRITIVAS ===")
df.describe()
```

```python
# Verificar valores únicos por coluna
print("=== VALORES ÚNICOS POR COLUNA ===")
for col in df.columns:
    unique_count = df[col].nunique()
    print(f"  {col}: {unique_count:,} valores únicos")
```

```python
# Verificar valores nulos
print("=== VALORES NULOS ===")
nulos = df.isnull().sum()
if nulos.sum() > 0:
    print(nulos[nulos > 0])
else:
    print("  ✓ Nenhum valor nulo encontrado!")
```

## Análise de Período Temporal

```python
print("=== ANÁLISE DE PERÍODO TEMPORAL ===")
if 'CO_ANO' in df.columns and 'CO_MES' in df.columns:
    anos = df['CO_ANO'].sort_values().unique()
    meses = df['CO_MES'].sort_values().unique()
    print(f"  Anos disponíveis: {list(anos)}")
    print(f"  Meses disponíveis: {list(meses)}")
    print(f"  Período: {min(anos)} a {max(anos)}")
else:
    print("  Colunas de tempo não encontradas!")
```

## Salvar Dados Processados

```python
# Salvar os dados carregados para uso na próxima etapa
output_file = os.path.join(OUTPUT_DIR, 'dados_carregados.csv')
df.to_csv(output_file, index=False)

print(f"✓ Dados salvos em: {output_file}")
print(f"  Total de registros: {len(df):,}")
print(f"  Tamanho do arquivo: {os.path.getsize(output_file)/(1024*1024):.2f} MB")
```
