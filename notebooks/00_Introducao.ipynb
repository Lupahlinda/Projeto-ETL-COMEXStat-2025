# Pipeline ETL - COMEX Stat 2025

## Visão Geral do Projeto

Este projeto implementa um pipeline ETL (Extract, Transform, Load) completo para processar dados de exportação brasileira do COMEX Stat 2025. O pipeline gera modelos de dados dimensionais (Star Schema) e aplica técnicas de machine learning para análise preditiva.

### Tecnologias Utilizadas
- **Python**: Linguagem principal
- **Pandas**: Manipulação de dados
- **Scikit-learn**: Machine learning (regressão linear)
- **Matplotlib**: Visualização de dados
- **Jupyter Notebook**: Ambiente de desenvolvimento interativo

### Estrutura dos Notebooks

1. **00_Introducao.ipynb** (este notebook): Visão geral e configuração
2. **01_Extract.ipynb**: Carregamento de dados brutos
3. **02_Transform.ipynb**: Transformação e enriquecimento de dados
4. **03_Regressao.ipynb**: Modelagem preditiva com regressão linear
5. **04_Load.ipynb**: Geração de modelos de dados (SQL e HTML[opcional])

### Fluxo do Pipeline

```
Dados Brutos (CSV) → Extract → Transform → Regressão → Load → Modelos de Dados
```

## Configuração Inicial

````python
# Importar bibliotecas principais
import pandas as pd
import numpy as np
import os
from datetime import datetime
import sys

# Adicionar diretório raiz ao path para importar módulos
project_root = os.path.dirname(os.path.dirname(os.path.abspath('__file__')))
sys.path.insert(0, project_root)

# Configurar display do pandas
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 50)
pd.set_option('display.float_format', '{:.2f}'.format)

# Diretórios do projeto
INPUT_DIR = '../data/input'
OUTPUT_DIR = '../data/output'
DICT_DIR = '../data/dictionaries'
SRC_DIR = 'src'

# Criar diretórios se não existirem
os.makedirs(INPUT_DIR, exist_ok=True)
os.makedirs(OUTPUT_DIR, exist_ok=True)
os.makedirs(DICT_DIR, exist_ok=True)

print("✓ Configuração inicial concluída!")
print(f"✓ Diretórios verificados: {INPUT_DIR}, {OUTPUT_DIR}, {DICT_DIR}")
````

## Verificar Estrutura do Projeto

````python
# Verificar arquivos de entrada
print("=== ARQUIVOS DE ENTRADA ===")
if os.path.exists(INPUT_DIR):
    arquivos_input = os.listdir(INPUT_DIR)
    for arquivo in arquivos_input:
        caminho = os.path.join(INPUT_DIR, arquivo)
        tamanho = os.path.getsize(caminho) / (1024*1024)  # MB
        print(f"  • {arquivo}: {tamanho:.2f} MB")
else:
    print("  Diretório de entrada não encontrado!")

print("\n=== ARQUIVOS DE DICIONÁRIO ===")
if os.path.exists(DICT_DIR):
    arquivos_dict = os.listdir(DICT_DIR)
    for arquivo in arquivos_dict:
        print(f"  • {arquivo}")
else:
    print("  Diretório de dicionários não encontrado!")
````
